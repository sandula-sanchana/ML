```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
data=load_breast_cancer()

X=data.data
Y=data.target

x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=42)

print(x_train.shape)
```

```python
scaler=StandardScaler()

x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test) #trasform only cause we already found the devieant and mean part from fit

```

```python
#pca=PCA(n_components=30)

#pca_train_data=pca.fit_transform(x_train_scaled)
#pca_test_data=pca.fit_transform(x_test_scaled)

pca_90=PCA(n_components=0.9) #we ask for 90% variance

pca_train_data=pca_90.fit_transform(x_train_scaled)
pca_test_data=pca_90.transform(x_test_scaled)

print(f"PCA Training Shape: {pca_train_data.shape} (We kept exactly {pca_90.n_components_} components!)")


```

```python
model=LogisticRegression()

model.fit(pca_train_data,y_train)

predictions=model.predict(pca_test_data)

accuracy=accuracy_score(y_test,predictions)

print(f"Model Accuracy using only {pca_90.n_components_} components: {accuracy * 100:.2f}%")

```

```python
new_data=X[0].reshape(1,-1)

new_data_scaled=scaler.transform(new_data)

new_data_scaled_pca=pca_90.transform(new_data_scaled)


new_predict=model.predict(new_data_scaled_pca)

# The target classes are 0 for Malignant, 1 for Benign
result = "Malignant" if new_predict[0] == 0 else "Benign"

print(f"3. Model Prediction: The tumor is {result}")


print(new_predict)
```
